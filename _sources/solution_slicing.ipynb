We can then load the data and visualize it. For describing that we want to _scroll_ through it, we can also use the term _slice_ through it, how Python developers call this action.

```python
%%bob
Load data/Haase_MRT_tfl3d1.tif and let me slice through it
```

```python
from skimage.io import imread
import stackview

image = imread("data/Haase_MRT_tfl3d1.tif")

stackview.slice(image)
```

Next, we want to generate code to transform the data so that we can slice through it from a different direction. The technical term for this transformation is _axis transposition_.

```python
%%bob
Transpose the data so that I can slice from the left to the right
```

```python
image_x = image.transpose(2, 1, 0)
stackview.slice(image_x)
```
