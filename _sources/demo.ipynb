# AI-assisted code-generation for image data analysis

`bia_bob` is a Python library that connects Jupyter with large language models (LLM) and custom bio-image analysis instructions.

When using bia-bob to prompt LLMs, the prompts you enter including existing variable names etc. are sent to the LLM provider. If you are using commercial cloud providers, where it is not obvious, what they are doing with data you send there, do not submit any sensitive, secret or private information. Also check bia-bob's code recommendations before executing them.

Also LLMs make mistakes. Review code carefully.

## Prompting
You can prompt for a single short task using `%bob`.

```python
%bob print hello world.
```

More advanced multi-line tasks can be prompted using `%%bob`.

```python
%%bob 
load data/blobs.tif and show it
```
