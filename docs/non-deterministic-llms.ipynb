# LLMs are non-deterministic

In this notebook we will demonstrate that LLMs are non-deterministic. If you ask them multiple times, they may generate different output. A major challenge is nowadays to deal with this fact properly. We use the [blablador](https://helmholtz-blablador.fz-juelich.de/) infrastructure at the Research Center Jülich. Before you can access it, you need to create an API key as explained [on this page](https://sdlaml.pages.jsc.fz-juelich.de/ai/guides/blablador_api_access/).

```python
import os
import openai
openai.__version__
```

```python
api_key = os.getenv("BLABLADOR_API_KEY")
if api_key is None:
    print("BLABLADOR_API_KEY environment variable not found. If you don't have it yet, get one as explained on this page: https://sdlaml.pages.jsc.fz-juelich.de/ai/guides/blablador_api_access/")
    api_key = getpass("Please enter your BLABLADOR_API_KEY:")

def prompt_blablador(message:str, model="00 - Aleph Kolibri", api_key=api_key):
    """A prompt helper function that sends a message to Blablador (FZ Jülich)
    and returns only the text response.
    """
    import os
    
    # convert message in the right format if necessary
    if isinstance(message, str):
        message = [{"role": "user", "content": message}]
    
    # setup connection to the LLM
    client = openai.OpenAI(base_url = "https://api.blablador.fz-juelich.de/v1/", api_key=api_key)
    
    response = client.chat.completions.create(
        model=model,
        messages=message
    )
    
    # extract answer
    return response.choices[0].message.content
```

```python
results = []
for i in range(10):
    result = prompt_blablador("How many Rs are in strawberry? Answer with the number only.", 
                              model="20 - EVE-Instruct - Expert Earth Observation and Earth Science (ES) domains")
    results.append(result)
    print(result)

print(results)
```
